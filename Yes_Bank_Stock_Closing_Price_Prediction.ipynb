## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Hypothesis 1

- Null Hypothesis (H0): There is no significant relationship between Open stock price and Close stock price.

- Alternate Hypothesis (H1): There is a significant relationship between Open stock price and Close stock price.

Hypothesis 2

- Null Hypothesis (H0): There is no significant relationship between High stock price and Low stock price.

- Alternate Hypothesis (H1): There is a significant relationship between High stock price and Low stock price.

Hypothesis 3

- Null Hypothesis (H0): Closing stock prices are normally distributed.

- Alternate Hypothesis (H1): Closing stock prices are not normally distributed.

```python
# Importing statistical library

from scipy.stats import pearsonr
from scipy.stats import shapiro
```

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H0): There is no significant relationship between Open stock price and Close stock price.

Alternate Hypothesis (H1): There is a significant relationship between Open stock price and Close stock price.

#### 2. Perform an appropriate statistical test.

```python
from scipy.stats import pearsonr
```

```python
# Perform Statistical Test to obtain P-Value
# Performing Pearson Correlation Test

corr, p_value = pearsonr(df['Open'], df['Close'])

print("Correlation Coefficient:", corr)
print("P-value:", p_value)
```

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test has been performed to measure the linear relationship between Open stock price and Close stock price.

##### Why did you choose the specific statistical test?

Pearson Correlation Test is suitable because both Open and Close prices are continuous numerical variables, and the objective is to determine the strength and significance of their linear relationship.

**Conclusion**

The correlation coefficient value of 0.9780 indicates a very strong positive relationship between Open stock price and Close stock price.

Since the p-value is significantly less than 0.05, we reject the Null Hypothesis.

Therefore, there is a statistically significant relationship between Open stock price and Close stock price.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H0): There is no significant relationship between High stock price and Low stock price.

Alternate Hypothesis (H1): There is a significant relationship between High stock price and Low stock price.

#### 2. Perform an appropriate statistical test.

```python
# Perform Statistical Test to obtain P-Value
# Performing Pearson Correlation Test

corr, p_value = pearsonr(df['High'], df['Low'])

print("Correlation Coefficient:", corr)
print("P-value:", p_value)
```

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test has been performed to measure the linear relationship between High stock price and Low stock price.

##### Why did you choose the specific statistical test?

Pearson Correlation Test is suitable because both High and Low prices are continuous numerical variables, and the objective is to determine the strength and significance of their linear relationship.



**Conclusion**

The correlation coefficient value of 0.9834 indicates a very strong positive relationship between High stock price and Low stock price.

Since the p-value is significantly less than 0.05, we reject the Null Hypothesis.

Therefore, there is a statistically significant relationship between High stock price and Low stock price.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H0): Closing stock prices are normally distributed.

Alternate Hypothesis (H1): Closing stock prices are not normally distributed.

#### 2. Perform an appropriate statistical test.

```python
from scipy.stats import shapiro
```

```python
# Perform Statistical Test to obtain P-Value
# Performing Shapiro-Wilk Test

stat, p_value = shapiro(df['Close'])

print("Test Statistic:", stat)
print("P-value:", p_value)
```

##### Which statistical test have you done to obtain P-Value?

Shapiro-Wilk Test has been performed to check whether the Closing stock prices follow normal distribution.

##### Why did you choose the specific statistical test?

Shapiro-Wilk Test is suitable for testing normality of numerical data and helps determine whether the dataset follows a normal distribution.

**Conclusion**

The test statistic value of 0.8170 and the extremely small p-value indicate that the Closing stock prices do not follow a normal distribution.

Since the p-value is significantly less than 0.05, we reject the Null Hypothesis.

Therefore, the Closing stock prices are not normally distributed.

### ML Model - 1  LINEAR REGRESSION MODEL

```python
# ML Model - 1 Implementation

# Fit the Algorithm

# Predict on the model

# Importing Linear Regression model

from sklearn.linear_model import LinearRegression

# Creating model object

lr_model = LinearRegression()

# Fit the Algorithm

lr_model.fit(X_train, y_train)

# Predict on the model

y_pred_lr = lr_model.predict(X_test)
```

```python
# Visualizing evaluation Metric Score chart
# Evaluation Metrics

mae_lr = mean_absolute_error(y_test, y_pred_lr)

mse_lr = mean_squared_error(y_test, y_pred_lr)

rmse_lr = np.sqrt(mse_lr)

r2_lr = r2_score(y_test, y_pred_lr)

# Printing metrics

print("MAE :", mae_lr)

print("MSE :", mse_lr)

print("RMSE :", rmse_lr)

print("R2 Score :", r2_lr)

# Evaluation metric score chart

lr_results = pd.DataFrame({
    'Metric':['MAE','MSE','RMSE','R2 Score'],
    'Score':[mae_lr,mse_lr,rmse_lr,r2_lr]
})

lr_results
```
